```python
# Variables
savings_goal = 15000
monthly_savings = 500
month_of_raise = 8          # month of salary raise
raise_amount = 250

# Counter Initialization
months_required = 0
total_savings = 0

# Loop until you reach the savings goal
while total_savings <= savings_goal:
    # Check if the raise is in effect and adjust monthly savings
    if months_required == month_of_raise:
        monthly_savings += raise_amount
        print("I got a raise")

    # Update total savings
    total_savings += monthly_savings

    # Increment the months counter
    months_required += 1
    print("after ",months_required," month(s), I have saved ", total_savings)
    

print(f"You will need {months_required} months to reach your savings goal of €{savings_goal}.")
```

```python
# variables
initial_investment = 1000
capitalisation_goal = 2000
interest_Rate = 0.1
years = 0

# 1) and 2) to reach investment goal, we will use a while loop to calculate the number of years it will take to reach the capitalisation goal
while initial_investment < capitalisation_goal:
    initial_investment = initial_investment * (1 + interest_Rate)
    years += 1
    print('my investment is worth', initial_investment, 'USD after', years, 'years')

print("it will take", years, "years to reach the capitalisation goal of", capitalisation_goal, "EUR.")

# 3) using a for look with multiple interest rates to calculate the number of years it will take to reach the capitalisation goal
rates = range(2,10)

for rates in range(2,11):
    interest_rate = rates / 100
    initial_investment = 1000
    years = 0
    while initial_investment < capitalisation_goal:
        initial_investment = initial_investment * (1 + interest_rate)
        #print(initial_investment)
        years += 1
    #print("with an interest rate of", interest_rate*100, "%", "it will take", years, "years to reach the capitalisation goal of", capitalisation_goal, "EUR.")
    print(f"At an annual interest rate of {interest_rate*100:.1f}% it will take {years} to reach the capitalisation goal of {capitalisation_goal} EUR.")
    
```
