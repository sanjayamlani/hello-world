```python
assets = ['bonds', 'stocks','crypto', 1, 1.8, True, False, [10,20]]
print(assets)
assets[0] = 'junk bonds'
print(assets)
assets.append('real estate')
print(assets)
assets.insert(1, 'commodities')
print(assets)
assets.remove('crypto')
print(assets)
del assets[2]
print(assets)
print(assets.pop())
print(assets*2)
```

```python
assets = ['bonds', 'stocks','crypto', 'real estate', 'commodities']
for i in range(len(assets)):
    print(assets[i])
    
```

```python
for element in assets:      # automatically iterates through the list and assigns each element to the variable 'element'
    print(element)
```
