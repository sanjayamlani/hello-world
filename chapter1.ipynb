```python
applestockprice = 135
s1=20
p1=60
s2=10
p2=145

#1
totalvalue = applestockprice * s1 + applestockprice * s2
print("$" + str(totalvalue))

#2
avpurchaseprice = (s1 * p1 + s2 * p2) / (s1 + s2)
print("$" + str(avpurchaseprice))

#3
totalpurchaseprice = s1 * p1 + s2 * p2

totalcapitalgain = totalvalue - totalpurchaseprice
print("$" + str(totalcapitalgain))

#4
FranceCGTRate = 0.3
BelguimCGTRate = 0.4    
BelguimCGTThreshold = 500

FranceCGT = totalcapitalgain * FranceCGTRate
BelguimCGT = max(totalcapitalgain - BelguimCGTThreshold,0) * BelguimCGTRate
print("$" + str(FranceCGT))
print("$" + str(BelguimCGT))
```
